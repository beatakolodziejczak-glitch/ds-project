```python
# Zadanie 1 - Tworzenie i indeksowanie listy

# Krok 1: Stworzenie listy z pomiarami temperatur
temperatury = [22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]

# Krok 2: Wyświetlenie temperatury z pierwszego dnia (indeks 0)
pierwszy_dzien = temperatury[0]
print(f"Temperatura z pierwszego dnia: {pierwszy_dzien}°C")

# Krok 3: Wyświetlenie temperatury z ostatniego dnia (indeks -1)
ostatni_dzien = temperatury[-1]
print(f"Temperatura z ostatniego dnia: {ostatni_dzien}°C")

# Krok 4: Wyświetlenie temperatur z trzech środkowych dni za pomocą wycinania (slicing)
# Indeksy trzech środkowych elementów to 2, 3 oraz 4, stąd zakres [2:5]
srodkowe_dni = temperatury[2:5]
print(f"Temperatury z trzech środkowych dni: {srodkowe_dni}")
```

```python
# Zadanie 2 - Dodawanie elementów do listy

# Krok 1: Zacznij od pustej listy
results = []

# Krok 2: Dodaj 5 wartości accuracy przy użyciu metody append
results.append(0.75)
results.append(0.82)
results.append(0.79)
results.append(0.85)
results.append(0.88)

# Krok 3: Wyświetlenie długości listy (len)
dlugosc = len(results)
print(f"Długość listy: {dlugosc}")

# Krok 4: Wyświetlenie średniej wartości (sum / len)
srednia = sum(results) / len(results)
print(f"Średnia wartość accuracy: {srednia:.4f}")

# Krok 5: Wyświetlenie maksymalnej wartości (max)
maksymalna = max(results)
print(f"Maksymalna wartość: {maksymalna}")

# Krok 6: Wyświetlenie minimalnej wartości (min)
minimalna = min(results)
print(f"Minimalna wartość: {minimalna}")
```

```python
# Zadanie 3 - Sortowanie listy i top-N

# Krok 1: Definicja listy wyników
wyniki = [0.78, 0.92, 0.85, 0.88, 0.81, 0.95, 0.79]

# Krok 2: Sortowanie listy malejąco przy użyciu .sort(reverse=True)
wyniki.sort(reverse=True)

# Krok 3: Wyświetlenie posortowanej listy
print(f"Posortowana lista (od najwyższego): {wyniki}")

# Krok 4: Wycięcie i wyświetlenie 3 najlepszych wyników za pomocą [:3]
top_3 = wyniki[:3]
print(f"3 najlepsze wyniki: {top_3}")
```

```python
# Zadanie 4 - Czyszczenie pojedynczej nazwy kolumny

# Krok 1: Definicja zmiennej z brudną nazwą
raw_col = "  Customer ID  "

# Krok 2: Usunięcie białych znaków, zmiana na małe litery i zastąpienie spacji podkreślnikami
# Możemy to zrobić krok po kroku (łańcuchowo)
clean_col = raw_col.strip().lower().replace(" ", "_")

# Krok 3: Wyświetlenie gotowego wyniku
print(f"Wynik czyszczenia: '{clean_col}'")
```

```python
# Zadanie 5 - Parsowanie linii CSV

# Krok 1: Definicja ciągu tekstowego z danymi
linia = "Alice,28,Data Scientist,75000"

# Krok 2: Rozdzielenie tekstu po przecinkach
elementy = linia.split(',')

# Krok 3: Wyciągnięcie wartości przy użyciu indeksów i konwersja typów danych
imie = elementy[0]                  # str
wiek = int(elementy[1])             # konwersja do int
zawod = elementy[2]                 # str
pensja = int(elementy[3])           # konwersja do int

# Krok 4: Wyświetlenie sformatowanego raportu zgodnie z wytycznymi
print(f"{imie} ma {wiek} lat, pracuje jako {zawod} i zarabia ${pensja}")
```

```python
# Zadanie 6 - Formatowanie raportu

# Krok 1: Definicja zmiennych z danymi
model_name = "XGBoost"
accuracy = 0.8934
training_time = 45.67

# Krok 2: Stworzenie wieloliniowego raportu przy użyciu f-stringa i potrójnych cudzysłowów
raport = f"""Model: {model_name}
  Accuracy: {accuracy:.2%}
  Training time: {training_time}s"""

# Krok 3: Wyświetlenie gotowego raportu
print(raport)
```

```python
# Zadanie 7 - Zliczanie wystąpień

# Krok 1: Definicja listy predykcji
predykcje = ['spam', 'ham', 'spam', 'spam', 'ham', 'ham', 'spam']

# Krok 2: Użycie metody .count() do policzenia wystąpień każdego słowa
liczba_spam = predykcje.count('spam')
liczba_ham = predykcje.count('ham')

# Krok 3: Wyświetlenie wyniku w formacie "spam: X, ham: Y"
print(f"spam: {liczba_spam}, ham: {liczba_ham}")
```

```python
# Zadanie 8 - Ekstrakcja informacji ze ścieżki

# Krok 1: Definicja zmiennej z nazwą pliku
filename = "model_results_2024_Q3.csv"

# Krok 2: Wyciągnięcie roku i kwartału za pomocą metody split
# Rozdzielamy tekst po podkreślnikach '_'
czesci = filename.split('_')

# Indeks 2 to rok '2024', a indeks 3 to 'Q3.csv', z którego musimy odciąć rozszerzenie (.split na kropce)
rok = czesci[2]
kwartal = czesci[3].split('.')[0]

print(f"Rok: {rok}")
print(f"Kwartał: {kwartal}")

# Krok 3: Sprawdzenie rozszerzenia za pomocą endswith()
czy_csv = filename.endswith('.csv')
print(f"Czy plik kończy się na .csv? {czy_csv}")
```

```python
# Zadanie 9 - Confusion matrix i metryki

# Krok 1: Definicja macierzy pomyłek
cm = [
    [90, 10],  # TN, FP (wiersz 0) - Dodano brakującą pierwszą linię
    [5, 45]    # FN, TP (wiersz 1)
]

# Krok 2: Dwuwymiarowe indeksowanie - wyciąganie wartości
tn = cm[0][0]
fp = cm[0][1]
fn = cm[1][0]
tp = cm[1][1]

# Krok 3: Obliczenie metryk zgodnie ze wzorami
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp)
recall = tp / (tp + fn)

# Krok 4: Wyświetlenie wyników sformatowanych jako procenty (z 2 miejscami po przecinku)
print(f"Accuracy: {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
```

```python
# Zadanie 14 - Ekstrakcja metadanych z nazwy pliku

# Krok 1: Definicja ścieżki wejściowej
sciezka = "/data/processed/2024/experiment_42_v3.csv"

# Krok 2: Rozdzielenie ścieżki za pomocą split na ukośnikach '/'
czesci = sciezka.split('/')

# Krok 3: Wyciąganie poszczególnych elementów przy użyciu indeksów i metod stringów
# Ostatni element na liście to nazwa pliku
nazwa_pliku = czesci[-1]

# Katalog to wszystko przed nazwą pliku (łączymy elementy od 0 do 4 za pomocą '/')
katalog = "/".join(czesci[:-1])

# Nazwę bez rozszerzenia uzyskujemy, odcinając ostatnie 4 znaki (".csv") za pomocą slicingu
nazwa_bez_rozszerzenia = nazwa_pliku[:-4]

# Rozszerzenie to po prostu ostatnie 4 znaki nazwy pliku
rozszerzenie = nazwa_pliku[-4:]

# Rok znajduje się na przedostatniej pozycji w pociętej ścieżce
rok = czesci[-2]

# Krok 4: Wyświetlenie czytelnego raportu z wyrównaniem (f-string)
print(f"{'Nazwa pliku:':<25} {nazwa_pliku}")
print(f"{'Katalog:':<25} {katalog}")
print(f"{'Nazwa bez rozszerzenia:':<25} {nazwa_bez_rozszerzenia}")
print(f"{'Rozszerzenie:':<25} {rozszerzenie}")
print(f"{'Rok z katalogu:':<25} {rok}")
```
