```python
# Utworzenie słownika metryki
metryki = {
    'accuracy': 0.94,
    'precision': 0.91,
    'recall': 0.89,
    'f1_score': 0.90
}

# 1. Wypisz najwyższą i najniższą wartość metryki
# Pobieramy same wartości słownika za pomocą .values()
wartosci = metryki.values()
print(f"Najwyższa wartość: {max(wartosci)}")
print(f"Najniższa wartość: {min(wartosci)}")

# 2. Wypisz średnią metrykę z dokładnością do 3 miejsc po przecinku
# Średnia to suma wszystkich elementów podzielona przez ich liczbę
srednia = sum(wartosci) / len(wartosci)
print(f"Średnia metryka: {srednia:.3f}")

# 3. Wypisz klucze posortowane alfabetycznie (sorted)
# Funkcja sorted() domyślnie sortuje klucze słownika
klucze_posortowane = sorted(metryki)
print(klucze_posortowane)

# 4. Wypisz wartości posortowane rosnąco
# Przekazujemy widok wartości do funkcji sorted()
wartosci_posortowane = sorted(wartosci)
print(wartosci_posortowane)
```

```python
# Dane początkowe
jezyki = ['Python', 'R', 'SQL', 'Python', 'Julia', 'R', 'Python', 'Scala', 'SQL']

# 1. Utwórz zbiór unikalne z tej listy
unikalne = set(jezyki)

# 2. Wypisz liczbę unikalnych języków
print(len(unikalne))

# 3. Dodaj do zbioru 'Go' i 'Rust' (użyj .update())
# Do metody .update() przekazujemy nową listę lub zbiór elementów
unikalne.update(['Go', 'Rust'])

# 4. Sprawdź (operatorem in), czy w zbiorze są: 'Python', 'Fortran', 'Julia'
print('Python' in unikalne)
print('Fortran' in unikalne)
print('Julia' in unikalne)

# 5. Usuń 'Scala' metodą .discard() (bezpieczną, bez błędu gdyby go nie było)
unikalne.discard('Scala')

# 6. Wypisz finalny zbiór oraz jego długość
print(unikalne)
print(len(unikalne))
```

```python
# Dane początkowe
grupa_a = {'python', 'pandas', 'numpy', 'sql', 'tableau'}
grupa_b = {'python', 'r', 'sql', 'ggplot2', 'dplyr'}

# 1. Suma zbiorów (wszystkie narzędzia)
print("1. Suma zbiorów:")
print(grupa_a | grupa_b)               # Operator |
print(grupa_a.union(grupa_b))          # Metoda .union()

# 2. Wspólne narzędzia (iloczyn)
print("\n2. Wspólne narzędzia:")
print(grupa_a & grupa_b)               # Operator &
print(grupa_a.intersection(grupa_b))   # Metoda .intersection()

# 3. Narzędzia tylko w grupie A (różnica grupa_a - grupa_b)
print("\n3. Narzędzia tylko w grupie A:")
print(grupa_a - grupa_b)               # Operator -
print(grupa_a.difference(grupa_b))     # Metoda .difference()

# 4. Narzędzia tylko w jednej z grup (różnica symetryczna)
print("\n4. Różnica symetryczna:")
print(grupa_a ^ grupa_b)               # Operator ^
print(grupa_a.symmetric_difference(grupa_b)) # Metoda .symmetric_difference()

# 5. Czy zbiory są rozłączne? (.isdisjoint())
print("\n5. Czy zbiory są rozłączne?")
print(grupa_a.isdisjoint(grupa_b))     # Zwróci False, bo mają wspólne elementy ('python', 'sql')
```
