```python
temperatury = [36.5, 37.8, 36.2, 38.5, 36.9, 39.2, 37.0]
liczba_goraczek = 0

for t in temperatury:
    if t >= 38.0:
        print(f"gorączka: {t}°C")
        liczba_goraczek += 1

lacznie_pomiarow = len(temperatury)
odsetek = (liczba_goraczek / lacznie_pomiarow) * 100

print(f"Odsetek przypadków z gorączką: {odsetek:.2f}%")
```

```python
trening = [(1.5, 'A'), (2.3, 'A'), (3.8, 'B'), (4.2, 'B'), (5.1, 'B')]
x_nowy = 3.5

# 1. Zbudowanie list comprehension z parami (odleglosc, klasa)
odleglosci = [(abs(x_nowy - x), klasa) for x, klasa in trening]

# 2. Posortowanie wyniku rosnąco
odleglosci_posortowane = sorted(odleglosci)

# 3. Wybranie k = 3 pierwszych elementów
sasiedzi = odleglosci_posortowane[:3]

# 4. Policzenie wystąpienia klas w sąsiadach za pomocą list comprehension z sum
liczba_A = sum(1 for _, klasa in sasiedzi if klasa == 'A')
liczba_B = sum(1 for _, klasa in sasiedzi if klasa == 'B')

# 5. Wyznaczenie przewidywanej klasy i wypisanie wyników
przewidywana_klasa = 'A' if liczba_A > liczba_B else 'B'

print("3 najbliższe punkty:")
for odl, kl in sasiedzi:
    print(f"Odległość: {odl:.1f}, Klasa: {kl}")
print(f"Przewidywana klasa: {przewidywana_klasa}")
```

```python
wyniki = [78, 82, 85, 88, 90, 92, 95, 97, 99, 45, 101, 103, 105, 180]

# 1. Posortowanie listy danych
posort = sorted(wyniki)

# 2. Wyznaczenie Q1 i Q3 "ręcznie" zgodnie ze specyfikacją zadania
Q1 = posort[len(posort) // 4]
Q3 = posort[3 * len(posort) // 4]

# 3. Obliczenie IQR oraz granic akceptowalnego przedziału wartości
IQR = Q3 - Q1
dolna = Q1 - 1.5 * IQR
gorna = Q3 + 1.5 * IQR
granice = [dolna, gorna]

# 4. Wyrażenie listowe (list comprehension) filtrujące wartości odstające
outliers = [x for x in wyniki if x < dolna or x > gorna]

# 5. Wypisanie wyników końcowych
print(f"Q1 = {Q1}")
print(f"Q3 = {Q3}")
print(f"IQR = {IQR}")
print(f"Granice = {granice}")
print(f"Outliers = {outliers}")
```
