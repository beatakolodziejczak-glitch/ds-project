```python
class Produkt:
    def __init__(self, nazwa: str, cena: float, ilosc_w_magazynie: int):
        self.nazwa = nazwa
        self.cena = cena
        self.ilosc_w_magazynie = ilosc_w_magazynie

    def pokaz_info(self):
        """Wyświetla informacje o produkcie"""
        print(f"Produkt: {self.nazwa:15s} | Cena: {self.cena:6.2f} zł | Stan magazynowy: {self.ilosc_w_magazynie} szt.")

    def sprzedaj(self, ilosc: int):
        """Zmniejsza ilosc_w_magazynie o podaną wartość, jeśli jest wystarczająca ilość"""
        if ilosc <= 0:
            print("Ilość sprzedawanych produktów musi być większa od zera.")
            return

        if self.ilosc_w_magazynie >= ilosc:
            self.ilosc_w_magazynie -= ilosc
            print(f"--> Sprzedano {ilosc} szt. produktu: {self.nazwa}")
        else:
            print(f"X Brak wystarczającej ilości produktu {self.nazwa} w magazynie (Dostępne: {self.ilosc_w_magazynie} szt.)")

    def dostawa(self, ilosc: int):
        """Zwiększa ilosc_w_magazynie o podaną wartość"""
        if ilosc > 0:
            self.ilosc_w_magazynie += ilosc
            print(f"<-- Przyjęto dostawę: {ilosc} szt. produktu: {self.nazwa}")
        else:
            print("Ilość dostawy must być większa od zera.")

# --- TESTOWANIE KLASY (3 OBIEKTY) ---

print("=== 1. TWORZENIE PRODUKTÓW ===")
p1 = Produkt("Laptop", 3499.99, 10)
p2 = Produkt("Mysz bezprzewodowa", 129.50, 50)
p3 = Produkt("Klawiatura mech.", 299.00, 5)

p1.pokaz_info()
p2.pokaz_info()
p3.pokaz_info()

print("\n=== 2. TEST METODY sprzedaj() ===")
p1.sprzedaj(3)   # Udana sprzedaż
p3.sprzedaj(10)  # Próba sprzedaży większej ilości niż w magazynie (komunikat błędu)
p1.pokaz_info()  # Sprawdzenie nowego stanu dla Laptopa (powinno być 7)

print("\n=== 3. TEST METODY dostawa() ===")
p3.dostawa(15)   # Zwiększenie stanu klawiatury
p3.pokaz_info()  # Sprawdzenie nowego stanu dla Klawiatury (powinno być 5 - 0 + 15 = 20)
```

```python
class Kalkulator:
    def dodaj(self, a: float, b: float) -> float:
        """Zwraca sumę a i b"""
        return a + b

    def odejmij(self, a: float, b: float) -> float:
        """Zwraca różnicę a i b"""
        return a - b

    def pomnoz(self, a: float, b: float) -> float:
        """Zwraca iloczyn a i b"""
        return a * b

    def podziel(self, a: float, b: float):
        """Zwraca iloraz a i b. Zawiera zabezpieczenie przed dzieleniem przez zero."""
        if b == 0:
            print("Błąd: Nie można dzielić przez zero!")
            return None
        return a / b

# --- TESTOWANIE KLASY ---

# Tworzenie obiektu
calc = Kalkulator()

# Przykładowe obliczenia
x, y = 12, 4

print("=== Przykładowe obliczenia ===")
print(f"{x} + {y} = {calc.dodaj(x, y)}")
print(f"{x} - {y} = {calc.odejmij(x, y)}")
print(f"{x} * {y} = {calc.pomnoz(x, y)}")
print(f"{x} / {y} = {calc.podziel(x, y)}")

print("\n=== Test dzielenia przez zero ===")
calc.podziel(10, 0)
```

```python
class BankAccount:
    def __init__(self, wlasciciel: str, saldo: float = 0.0):
        self.wlasciciel = wlasciciel
        self.saldo = saldo

    def wplac(self, kwota: float):
        """Zwiększa saldo o podaną kwotę"""
        if kwota > 0:
            self.saldo += kwota
            print(f"[{self.wlasciciel}]: Wpłacono {kwota:.2f} zł.")
        else:
            print("Kwota wpłaty musi być większa od zera.")

    def wyplac(self, kwota: float):
        """Zmniejsza saldo o podaną kwotę (tylko jeśli wystarczy środków)"""
        if kwota <= 0:
            print("Kwota wypłaty musi być większa od zera.")
            return

        if self.saldo >= kwota:
            self.saldo -= kwota
            print(f"[{self.wlasciciel}]: Wypłacono {kwota:.2f} zł.")
        else:
            print(f"X [{self.wlasciciel}]: Odmowa operacji. Brak środków na koncie (Dostępne: {self.saldo:.2f} zł, próba wypłaty: {kwota:.2f} zł).")

    def pokaz_saldo(self):
        """Wyświetla aktualne saldo konta"""
        print(f"Konto: {self.wlasciciel:12s} | Saldo: {self.saldo:.2f} zł")


# --- TESTOWANIE KLASY (KILKA KONT) ---

print("=== 1. TWORZENIE KONT ===")
konto_jan = BankAccount("Jan Kowalski", 1000.0)
konto_anna = BankAccount("Anna Nowak")  # Domyślne saldo: 0.0

konto_jan.pokaz_saldo()
konto_anna.pokaz_saldo()

print("\n=== 2. TEST METODY wplac() ===")
konto_anna.wplac(500.50)
konto_anna.pokaz_saldo()

print("\n=== 3. TEST METODY wyplac() ===")
# Udana wypłata u Jana
konto_jan.wyplac(400.0)
konto_jan.pokaz_saldo()

# Próba wypłaty ponad stan u Anny (zabezpieczenie)
konto_anna.wyplac(600.0)
konto_anna.pokaz_saldo()
```

```python
class Ksiazka:
    def __init__(self, tytul: str, autor: str, rok_wydania: int, liczba_stron: int):
        self.tytul = tytul
        self.autor = autor
        self.rok_wydania = rok_wydania
        self.liczba_stron = liczba_stron

    def __str__(self) -> str:
        """Zwraca czytelną reprezentację tekstową książki"""
        return f"'{self.tytul}' – {self.autor} ({self.rok_wydania}), {self.liczba_stron} str."

# --- TWORZENIE LISTY 5 KSIĄŻEK I WYŚWIETLANIE ---

# 1. Tworzenie listy obiektów
biblioteka = [
    Ksiazka("Wiedźmin: Ostatnie życzenie", "Andrzej Sapkowski", 1993, 288),
    Ksiazka("Heban", "Ryszard Kapuściński", 1998, 328),
    Ksiazka("Pan Tadeusz", "Adam Mickiewicz", 1834, 380),
    Ksiazka("Solaris", "Stanisław Lem", 1961, 240),
    Ksiazka("Bieguni", "Olga Tokarczuk", 2007, 456)
]

# 2. Wyświetlanie przy użyciu print() w pętli
print("=== Moja Biblioteka (Metoda __str__) ===")
for ksiazka in biblioteka:
    print(ksiazka)
```

```python
class Student:
    def __init__(self, imie: str, numer_indeksu: str, oceny: list = None):
        self.imie = imie
        self.numer_indeksu = numer_indeksu
        # Bezpieczne inicjalizowanie pustej listy dla ocen
        self.oceny = oceny if oceny is not None else []

    def dodaj_ocene(self, ocena: float):
        """Dodaje ocenę do listy ocen studenta"""
        if 2.0 <= ocena <= 5.5:  # Opcjonalne sprawdzenie poprawności skali ocen
            self.oceny.append(ocena)
        else:
            print(f"Niepoprawna ocena: {ocena}. Wprowadź wartość z zakresu 2.0 - 5.5.")

    def srednia(self) -> float:
        """Zwraca średnią ocen studenta. Jeśli brak ocen, zwraca 0.0"""
        if not self.oceny:
            return 0.0
        return sum(self.oceny) / len(self.oceny)

    def __str__(self) -> str:
        """Reprezentacja tekstowa studenta"""
        return f"Student: {self.imie:12s} | Indeks: {self.numer_indeksu} | Oceny: {self.oceny}"


# --- TWORZENIE 3 STUDENTÓW, DODAWANIE OCEN I TESTOWANIE ---

print("=== 1. TWORZENIE STUDENTÓW ===")
s1 = Student("Anna", "123456")
s2 = Student("Jan", "234567", [4.0, 5.0])
s3 = Student("Katarzyna", "345678")

print(s1)
print(s2)
print(s3)

print("\n=== 2. DODAWANIE OCEN ===")
s1.dodaj_ocene(4.5)
s1.dodaj_ocene(3.5)
s1.dodaj_ocene(5.0)

s2.dodaj_ocene(3.0)

s3.dodaj_ocene(5.0)
s3.dodaj_ocene(5.0)

print(s1)
print(s2)
print(s3)

print("\n=== 3. WYŚWIETLANIE ŚREDNICH ===")
print(f"{s1.imie}: średnia = {s1.srednia():.2f}")
print(f"{s2.imie}: średnia = {s2.srednia():.2f}")
print(f"{s3.imie}: średnia = {s3.srednia():.2f}")
```

```python
class Pojazd:
    def __init__(self, marka: str, model: str, rok: int):
        self.marka = marka
        self.model = model
        self.rok = rok

    def pokaz_info(self):
        """Wyświetla informacje o pojeździe"""
        print(f"Pojazd: {self.marka} {self.model} ({self.rok})")


class Samochod(Pojazd):
    def __init__(self, marka: str, model: str, rok: int, liczba_drzwi: int):
        # Wywołanie konstruktora klasy bazowej (Pojazd)
        super().__init__(marka, model, rok)
        self.liczba_drzwi = liczba_drzwi

    def pokaz_info(self):
        """Nadpisuje metodę pokaz_info, dodając informację o liczbie drzwi"""
        print(f"Samochód: {self.marka} {self.model} ({self.rok}) | Drzwi: {self.liczba_drzwi}")


# --- TWORZENIE OBIEKTU I WYWOŁANIE METODY ---

print("=== Test klasy potomnej Samochod ===")
# Stworzenie obiektu klasy Samochod
moje_auto = Samochod("Toyota", "Corolla", 2022, 5)

# Wywołanie nadpisanej metody pokaz_info()
moje_auto.pokaz_info()
```

```python
class Pracownik:
    def __init__(self, imie: str, stanowisko: str, pensja: float):
        self.imie = imie
        self.stanowisko = stanowisko
        self.pensja = pensja

    def podwyzka(self, procent: float):
        """Zwiększa pensję o podany procent (np. 10 dla 10%)"""
        if procent > 0:
            kwota_podwyzki = self.pensja * (procent / 100)
            self.pensja += kwota_podwyzki
            print(f"--> Przyznano podwyżkę {procent}% dla pracownika: {self.imie}")
        else:
            print("Procent podwyżki musi być większy od zera.")

    def __str__(self) -> str:
        """Zwraca reprezentację tekstową pracownika"""
        return f"Pracownik: {self.imie:12s} | Stanowisko: {self.stanowisko:15s} | Pensja: {self.pensja:.2f} zł"


# --- TWORZENIE 3 PRACOWNIKÓW I DAWANIE PODWYŻEK ---

print("=== 1. TWORZENIE PRACOWNIKÓW ===")
p1 = Pracownik("Jan Kowalski", "Programista", 8000.0)
p2 = Pracownik("Anna Nowak", "Analityk", 7500.0)
p3 = Pracownik("Piotr Bekas", "Tester", 6000.0)

print(p1)
print(p2)
print(p3)

print("\n=== 2. NALICZANIE PODWYŻEK ===")
p1.podwyzka(10)  # 10% podwyżki dla Jana
p2.podwyzka(15)  # 15% podwyżki dla Anny
p3.podwyzka(5)   # 5% podwyżki dla Piotra

print("\n=== 3. STAN PO PODWYŻKACH ===")
print(p1)
print(p2)
print(p3)
```

```python
class Zadanie:
    def __init__(self, opis: str, status: str = "do zrobienia"):
        self.opis = opis
        self.status = status

    def oznacz_jako_zrobione(self):
        """Zmienia status zadania na 'zrobione'"""
        self.status = "zrobione"

    def __str__(self) -> str:
        """Zwraca czytelną reprezentację tekstową zadania"""
        # Wizualny znacznik statusu dla lepszej czytelności w konsoli
        znacznik = "[X]" if self.status == "zrobione" else "[ ]"
        return f"{znacznik} Zadanie: {self.opis:35s} | Status: {self.status}"


# --- TWORZENIE LISTY 5 ZADAŃ I ICH AKTUALIZACJA ---

print("=== 1. TWORZENIE LISTY TODO ===")
lista_todo = [
    Zadanie("Zrobić zakupy spożywcze"),
    Zadanie("Uruchomić kod w Google Colab"),
    Zadanie("Przesłać zmiany do repozytorium"),
    Zadanie("Nagrać materiał wideo z lekcji"),
    Zadanie("Odpocząć po nauce")
]

# Wyświetlenie początkowej listy
for zadanie in lista_todo:
    print(zadanie)

print("\n=== 2. OZNACZANIE CZĘŚCI ZADAŃ JAKO ZROBIONE ===")
# Oznaczamy indeksy 1 oraz 2 jako zrobione
lista_todo[1].oznacz_jako_zrobione()
lista_todo[2].oznacz_jako_zrobione()

# Wyświetlenie zaktualizowanej listy
for zadanie in lista_todo:
    print(zadanie)
```

```python
class FiltrowanieDodatnich:
    def wykonaj(self, dane: list) -> list:
        """Usuwa liczby ujemne z listy (zostawia >= 0)"""
        return [x for x in dane if x >= 0]


class Zaokraglanie:
    def wykonaj(self, dane: list) -> list:
        """Zaokrągla elementy listy do liczb całkowitych"""
        return [round(x) for x in dane]


class Skalowanie:
    def __init__(self, wspolczynnik: float):
        self.wspolczynnik = wspolczynnik

    def wykonaj(self, dane: list) -> list:
        """Mnoży każdy element przez stały współczynnik"""
        return [x * self.wspolczynnik for x in dane]


class Sumowanie:
    def wykonaj(self, dane: list) -> float:
        """Zwraca sumę elementów listy"""
        return sum(dane)


class Pipeline:
    def __init__(self):
        # Inicjalizacja listy komponentów wchodzących w skład potoku danych
        self.kroki = []

    def dodaj_krok(self, komponent):
        """Dodaje kolejny komponent do pipeline'u"""
        self.kroki.append(komponent)

    def uruchom(self, dane_wejsciowe: list):
        """Przepuszcza dane sekwencyjnie przez wszystkie dodane kroki"""
        aktualne_dane = dane_wejsciowe.copy()

        print(f"Dane wejściowe: {aktualne_dane}")

        for krok in self.kroki:
            # Dynamiczne wykonywanie operacji niezależnie od typu komponentu
            aktualne_dane = krok.wykonaj(aktualne_dane)
            print(f" po kroku [{krok.__class__.__name__}]: {aktualne_dane}")

        return aktualne_dane


# --- TESTOWANIE PIPELINE'U ---

# 1. Przygotowanie danych z zadania
dane_testowe = [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]

# 2. Tworzenie obiektów komponentów
filtr = FiltrowanieDodatnich()
zaokraglanie = Zaokraglanie()
skalowanie = Skalowanie(wspolczynnik=2.0)  # Przykładowe mnożenie x2
sumowanie = Sumowanie()

# 3. Budowa i konfiguracja potoku (Pipeline)
potok = Pipeline()
potok.dodaj_krok(filtr)         # Krok 1: Filtrowanie ujemnych
potok.dodaj_krok(zaokraglanie)  # Krok 2: Zaokrąglanie do całkowitych
potok.dodaj_krok(skalowanie)     # Krok 3: Skalowanie (mnożenie przez stałą)
potok.dodaj_krok(sumowanie)     # Krok 4: Agregacja (sumowanie)

# 4. Uruchomienie całego procesu
print("=== Uruchomienie Pipeline'u Danych ===")
wynik_koncowy = potok.uruchom(dane_testowe)

print("\n=== Wynik Końcowy ===")
print(f"Suma końcowa: {wynik_koncowy}")
```

```python
class Akcja:
    def __init__(self, nazwa: str, cena_kupna: float, cena_aktualna: float):
        self.nazwa = nazwa
        self.cena_kupna = cena_kupna
        self.cena_aktualna = cena_aktualna


class Portfel:
    def __init__(self, kapital_poczatkowy: float, akcje=None):
        self.kapital_poczatkowy = kapital_poczatkowy
        # Przechowujemy jako listę par: [[obiekt_akcji, ilosc], ...]
        self.akcje = akcje if akcje is not None else []

    def kup_akcje(self, akcja: Akcja, ilosc: int):
        """Dodaje akcję i jej ilość do portfela"""
        self.akcje.append([akcja, ilosc])
        print(f"Kupiono: {ilosc}x {akcja.nazwa}")

    def sprzedaj_akcje(self, akcja: Akcja, ilosc: int):
        """Pomniejsza liczbę akcji w portfelu"""
        for element in self.akcje:
            if element[0].nazwa == akcja.nazwa:
                element[1] -= ilosc
                print(f"Sprzedano: {ilosc}x {akcja.nazwa}")
                if element[1] <= 0:
                    self.akcje.remove(element)
                break

    def aktualizuj_ceny(self, slownik_cen: dict):
        """Aktualizuje cenę aktualną na podstawie słownika"""
        for element in self.akcje:
            nazwa = element[0].nazwa
            if nazwa in slownik_cen:
                element[0].cena_aktualna = slownik_cen[nazwa]

    def wartosc_portfela(self) -> float:
        """Oblicza sumaryczną wartość akcji w portfelu"""
        return sum(element[1] * element[0].cena_aktualna for element in self.akcje)

    def zysk_strata(self) -> float:
        """Różnica między aktualną wartością a kapitałem początkowym"""
        return self.wartosc_portfela() - self.kapital_poczatkowy


# --- SYMULACJA ZADANIA (5 AKCJI) ---
baza = Portfel(kapital_poczatkowy=5000.0)

# Definicja 5 obiektów klas Akcja
akcja1 = Akcja("CDP", 100.0, 100.0)
akcja2 = Akcja("PKO", 40.0, 40.0)
akcja3 = Akcja("ALE", 35.0, 35.0)
akcja4 = Akcja("KGH", 120.0, 120.0)
akcja5 = Akcja("PZU", 30.0, 30.0)  # <-- Tutaj błąd został naprawiony!

print("--- Kupowanie akcji ---")
baza.kup_akcje(akcja1, 10)
baza.kup_akcje(akcja2, 20)
baza.kup_akcje(akcja3, 30)
baza.kup_akcje(akcja4, 5)
baza.kup_akcje(akcja5, 15)

print(f"\nWartość portfela po zakupie: {baza.wartosc_portfela():.2f} zł")

print("\n--- Zmiana cen rynkowych ---")
baza.aktualizuj_ceny({"CDP": 120.0, "PKO": 35.0, "ALE": 45.0, "KGH": 110.0, "PZU": 32.0})

print(f"Aktualna wartość portfela: {baza.wartosc_portfela():.2f} zł")
print(f"Zysk / Strata względem kapitału: {baza.zysk_strata():.2f} zł")

print("\n--- Sprzedaż części akcji ---")
baza.sprzedaj_akcje(akcja1, 5)

print(f"\nKońcowa wartość portfela: {baza.wartosc_portfela():.2f} zł")
```
